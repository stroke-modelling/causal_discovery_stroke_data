Extending and joining dataframes

```python
df_a=pd.DataFrame(
    [{"Name":"Dan","Trainer?":"Yes"},
     {"Name":"Sammi","Trainer?":"Yes"}]
)

df_b=pd.DataFrame(
    [{"Name":"Mike","Trainer?":"No"},
     {"Name":"Tom","Trainer?":"No"}]
)
```

```python
df_a
```

```python
df_b
```

```python
df_c=pd.concat([df_a,df_b],axis=0)
df_c
```

```python
#this set has duplicate index values so to reset this we use reset function
df_c.reset_index()
```

```python
#this has now become a new column so to avoid this let's use drop=True
df_c.reset_index(drop=True)
```
