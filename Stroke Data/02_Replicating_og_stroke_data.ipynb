Normal distribution for ages

```python
def round_ages(ages):
    """Round a np.array of ages to the nearest 5-year age band."""
    return ((ages+1e-5) // 5) * 5 + 2.5
```

```python
mean=80
standard_deviation=20
size=20000
np.random.seed(42)
a = np.random.normal(mean,standard_deviation,size)
a= np.clip(a,a_min=37.5, a_max=np.inf)
a[a>95]=np.nan
a=a[:10000]
a=round_ages(a)
plt.hist(a, 50)
plt.axvline(a.mean(), color='k', linestyle='dashed', linewidth=2)
plt.show()
```
