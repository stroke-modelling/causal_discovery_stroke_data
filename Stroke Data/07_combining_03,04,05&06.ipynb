### Stroke onset during sleep and Precise stroke onset known
* df_a- initial dataframe with data from one drive
* b= Stroke not happening during sleep + Precise onset not known
* c= Stroke happening during sleep + Precise onset not known
* d= Stroke not happening during sleep + Precise onset known
* e= Stroke happening during sleep + Precise onset known
* f1= Random choice of probability of stroke happening during sleep
* f2= Random choice of whether or not a patient has the precise onset known based on previous probabilities
* df_b= series of f1
* df_c= series of f2
* df_d= final dataframe with f1 and f2 combined

```python
df_a=pd.read_csv("combo_sleep_precise.csv", index_col="precise_onset_known")
```

```python
b=df_a.loc[0,"0"]#0.33568
c=df_a.loc[0,"1"]
d=df_a.loc[1,"0"]
e=df_a.loc[1,"1"]
#no quote marks for rows

b1=b/(b+d)
d1=d/(b+d)

c1=c/(c+e)
e1=e/(c+e)
```

```python
num_rows = 1000
p = c+e
np.random.seed(42)
f1 = np.random.binomial(1, p, num_rows)


f2=np.full(num_rows,42)
inds_f1_0 = np.where(f1 == 0)
num_f1_0 = len(inds_f1_0[0])

inds_f1_1= np.where(f1==1)
num_f1_1= len(inds_f1_1[0])

f2_vals = np.random.choice([0, 1], num_f1_0, p=[b1,d1])
f2[inds_f1_0] = f2_vals

f1_vals=np.random.choice([0, 1], num_f1_1, p=[c1,e1])
f2[inds_f1_1]=f1_vals
```

```python
df_b=pd.Series(f1)
df_c=pd.Series(f2)
```

```python
df_d=pd.concat([df_b,df_c], axis=1)
df_d.columns=["Onset_during_sleep","Precise_onset_known"]
df_d
```

### Age and Prior disability
* df_i- initial dataframe with data from one drive
* f5= Random choice of probability of patient having disability between 0 to 5
* f2= Random choice of patient age based on previous probabilities
* df_j= final dataframe with f1 and f2 combined

```python
df_i=pd.read_csv("combo_age_disability.csv", index_col="prior_disability")
```

```python
vals_f5 = df_i.columns
probs_f5 = df_i.sum(axis='rows')
f5 = np.random.choice(vals_f5, size=num_samples, p=probs_f5)
```

```python
f6 = np.full(len(f5), 42)
vals_f6 = df_i.index

for j in vals_f5:
    inds_j = np.where(f5 == j)
    num_j = len(inds_j[0])
    probs_j = df_i[j]
    probs_j = probs_j/ probs_j.sum()
    f6_j = np.random.choice(vals_f6, size=num_j, p=probs_j)
    f6[inds_j] = f6_j
    print(j, num_j, probs_j, f6_j)
```

```python
df_j=pd.DataFrame([f5, f6]).T
df_j.columns=["Patient age","Disabilities (0-5)"]
df_j
```

### Arrival to scan time and stroke teams

```python
df_k=pd.read_csv("arrival_to_scan_time_stats_by_stroke_team.csv", index_col="stats")
```

```python
df_l=pd.read_csv("stroke_team.csv")
n=df_l["stroke_team"]
np.random.seed(42)
stroke_team_number=np.random.choice(n,num_samples,p=df_l["count"])
```

```python
stn=pd.Series(stroke_team_number)
```

```python
team_names=df_k.columns
f9=np.full(num_samples,42)

for o in team_names:
    p=df_k.loc["mu_for_lognorm", o]
    q=df_k.loc["sigma_for_lognorm", o]

    if int(o) in stn.value_counts().index:
        r=stn.value_counts()[int(o)]
    else:
        r=0

    f8 = np.random.default_rng()
    n_gen = f8.lognormal(p, q, size=r)

    inds_f8_0 = np.where(stroke_team_number == int(o))
    f9[inds_f8_0] = n_gen
```

```python
df_m=pd.DataFrame([pp, f9]).T
df_m.columns=["Stroke team","Arrival to scan time"]
df_m
```
