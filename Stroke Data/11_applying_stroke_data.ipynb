Using previously generated data (Notebook 10) to calculate thrombolysis and outcome

```python
import pandas as pd
import numpy as np
import random as rn
import matplotlib.pyplot as plt
```

```python
num_samples=10000
```

```python
col_name={
    "Mild Stroke":"stroke_severity_mild",
    "Moderate Stroke":"stroke_severity_moderate",
    "Severe Stroke":"stroke_severity_severe",
    "Age":"age_80plus",
    "Utility":"prior_util",
    "AF_patients":"atrial_fibrillation",
    "AF_patients_on_anticoagulants":"afib_anticoagulant",
    "Onset_during_sleep":"onset_during_sleep",
    "Precise_onset_known":"precise_onset_known",
    "Arrival to scan time":"arrival_to_scan_time"}
df_data=pd.read_csv("final_10_08.csv", index_col=0)
df_data=df_data.rename(columns=col_name)
df_data
```

```python
# Make one-hot-encoded data:
df_data_ohe = pd.get_dummies(df_data["Stroke team"]).astype(int)
# Merge this into the rest of the iris data:
df_data = pd.concat((df_data, df_data_ohe), axis='columns')
df_data
```

```python
df_file=pd.read_csv("effects_linreg_allpatients_mrsleq2.csv", index_col=0)
df_file=df_file.fillna(0)
df_file
```

```python
stroke_team_names = df_file.index[df_file.index.str.isnumeric()].values
stroke_team_names
```

```python
df_data=df_data.rename(columns=dict(zip([int(t) for t in stroke_team_names], stroke_team_names)))
```

```python
df_data=df_data[(df_data[stroke_team_names] == 1).any(axis="columns")]
df_data
```

```python
patients=len(df_data)
patients
```

```python
data_columns=df_data.columns
data_columns
```

```python
file_columns=df_file.index
file_columns
```

```python
common_columns=[]

for i in data_columns:
    if i in file_columns:
        common_columns.append(i)
```

```python
common_columns
```

```python
noise_t = np.random.normal(0, 0.1, patients)

t = np.sum((
    [noise_t] +
    [df_data[col] * df_file.loc[col, 'thrombolysis'] for col in common_columns]
), axis=0)

import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def convert_to_binary(x, stochastic=True):
    p = sigmoid(x)
    if stochastic:
        return np.random.choice([0, 1], 1, p=[1 - p, p])
    else:
        return int(p > 0.5)

t_bin = np.array([convert_to_binary(x) for x in t]).flatten()
```

```python
df_data['thrombolysis'] = t_bin
df_data
```

```python
out_col="discharge_mrsleq2"

def outcome_calculator(out_col, df_data, df_file):

    np.random.seed(42)
    noise_out = np.random.normal(0, 0.1, patients)

    out = np.sum((
        [noise_out] +
        [df_data[col] * df_file.loc[col, out_col] for col in common_columns]
    ), axis=0) + df_file.loc['const', out_col]

    out_bin = np.array([sigmoid(x) for x in out]).flatten()
    return out_bin
```

```python
df_data[out_col]=outcome_calculator(out_col, df_data, df_file)
df_data
```

```python
df_data.to_csv("final_results_11_08.csv")
```
