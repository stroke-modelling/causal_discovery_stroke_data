# Example round

```python
def round_ages(ages):
    """Round a np.array of ages to the nearest 5-year age band."""
    return ((ages+1e-5) // 5) * 5 + 2.5

round_ages(np.array([40, 41, 43, 45, 46, 48, 51]))
```

```python
def round_times(t, r=5):
    """Round a np.array of times to their previous time interval r."""
    return ((t+1e-5) // r) * r

round_times(np.array([40, 41, 43, 45, 46, 48, 51]))
```
