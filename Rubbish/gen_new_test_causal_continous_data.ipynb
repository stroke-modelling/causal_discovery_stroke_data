We have generated data similar to causal test data except it has 10000 samples this time.

```python
import pandas as pd
import numpy as np
import random as rn
import matplotlib.pyplot as plt
```

```python
num_samples=10000
```

A- A

```python
np.random.seed(42)
A = np.random.normal(-0.25, 1, num_samples)
```

B- B

```python
np.random.seed(42)
B = np.random.normal(0.20, 1, num_samples)
```

C- C

```python
np.random.seed(42)
C = np.random.normal(-0.69, 1, num_samples)
```

D- D

```python
p = 0.73
np.random.seed(42)
D = np.random.binomial(1, p, num_samples)
```

E- E

```python
np.random.seed(42)
E= np.random.normal(0.90, 1, num_samples)
```

Making 1 big dataframe

```python
df=pd.DataFrame([A, B, C, D, E])
df.index=["A","B","C","D","E"]
df=df.T
```

```python
df.head()
```

Adding a yes or no value for treatment


Note: Formula is Treatment= Noise + 0.45A + 0.15E + 1.3D

We had previously used 4.5 but any value of variables above 3 disrupts the flow (based off of a mathematical thing), hence we have reduced it back to under 3

```python
np.random.seed(42)
noise_t = np.random.normal(0, 0.1, num_samples)
t = np.sum([noise_t] + [0.45*A] + [0.15*E] + [1.32*D], axis=0)
```

```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def convert_to_binary(x, stochastic=True):
    p = sigmoid(x)
    if stochastic:
        return np.random.choice([0, 1], 1, p=[1 - p, p])
    else:
        return int(p > 0.5)
```

```python
t_bin = np.array([convert_to_binary(x) for x in t]).flatten()
```

```python
df['Treatment'] = t_bin
```

```python
df
```

Adding Outcome values

Note: The right formula used here is Outcome= Noise + 5*Treatment + 0.33*A + 0.39*E + 0.02*B

We had previously used 3.3 and 3.9 but any value of variables above 3 disrupts the flow (based off of a mathematical thing), hence we have reduced it back to under 3

```python
np.random.seed(42)
noise_out = np.random.normal(0, 0.1, num_samples)
out = np.sum([noise_out] + [5*t_bin] + [0.33*A] + [0.39*E] + [0.02*B*t_bin], axis=0)
```

```python
noise_out
```

```python
df['Outcome'] = out
```
