## Machine Learning: Iris Dataset

### 1. Project Overview & Context
The Iris dataset is a classic benchmark in machine learning and pattern recognition, originally introduced by the statistician Ronald Fisher. It consists of 150 observations of Iris flowers across three distinct species:
* Setosa
* Versicolor
* Virginica

![Picture of the three species of irises](Iris_species.png)

For each flower, four physical measurements (features) are captured in centimeters:

* Sepal Length
* Sepal Width
* Petal Length
* Petal Width

In this project, we split the multi-class problem into three independent binary classifiers

* Setosa Classifier:  Predicts 1 if the flower is Setosa; 0 if it is any other species.
* Versicolor Classifier:  Predicts 1 if the flower is Versicolor; 0 if it is any other species.
* Virginica Classifier:  Predicts 1 if the flower is Virginica; 0 if it is any other species.

Our analysis is split across four distinct, sequential notebooks. 

* **Notebook 01a**: Simple Machine Learning Iris Classifier


Goal: Establish our baseline models using basic training and testing dataset splits.

Key Concept: Train-Test Splitting. We hold out 25% of our data to test how well our model generalizes to unseen data.

Outcome: A simple implementation showing perfect accuracy, which raises immediate questions about whether our model is robust or just lucky.

* **Notebook 02**: Rigorous Cross-Validation and Metric Expansion


Goal: Replace the volatile single-train-test split with a rigorous statistical evaluation.

Key Concept: Stratified K-Fold Cross-Validation. By splitting the dataset into 10 balanced folds, we guarantee that every single sample is used for validation. We also expand our evaluation beyond simple accuracy to include Precision, Recall, F1-Score, and Specificity.

Outcome: We reveal the true, realistic accuracy of our classifiers when exposed to balanced data slices, highlighting a minor classification overlap between Versicolor and Virginica.

* **Notebook 03**: Probabilistic Thresholds and ROC/AUC Analysis


Goal: Move away from rigid "binary" (0 or 1) predictions and analyze the continuous probabilities generated by our models.

Key Concept: Receiver Operating Characteristic (ROC) and Area Under the Curve (AUC). We map how our True Positive Rate and False Positive Rate balance against each other across every possible probability cutoff from 0.0 to 1.0.

Outcome: We construct threshold-agnostic profiles for our classifiers, finding that our models remain highly stable even at aggressive threshold limits.

* **Notebook 04**: Local Interpretability with Shapley (SHAP) Values


Goal: Open our "black-box" XGBoost models to explain exactly why individual predictions are made.

Key Concept: Shapley Additive exPlanations (SHAP). Using game theory, we assign a numeric contribution score to each physical measurements.

Outcome: We map global feature behaviors, local "waterfall" profiles for specific flowers, and plot exact feature-level relationships.
